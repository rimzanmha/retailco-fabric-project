### nb_load_dim
Loads **one** dimension CSV from `Files/bronze/` into a Delta table `silver_dim_<name>`.
Called inside the pipeline's **ForEach** loop (`@item().name`), so one notebook serves store, product and customer.

```python
# parameters (the pipeline overrides these)
dim_name = "customer"          # store | product | customer
run_date = "2026-11-08"
```

```python
from pyspark.sql import functions as F

key = f"{dim_name}_id"

df = (spark.read.option("header", True).option("inferSchema", True)
      .csv(f"Files/bronze/dim_{dim_name}.csv"))

df = df.dropna(subset=[key]).dropDuplicates([key])
df = df.withColumn("_loaded_at", F.current_timestamp())

df.write.mode("overwrite").format("delta").saveAsTable(f"silver_dim_{dim_name}")
print(dim_name, "loaded:", df.count(), "rows")
```
